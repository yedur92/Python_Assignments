### **Number Guessing Game**

**Problem Statement**: Create a Python program that implements a simple number guessing game using a while loop.<br> The program should make use of control statements such as else, break, and continue.

```python
import random
secret_number = random.randint(1, 10)

attempts = 3

while attempts > 0:
    guess = int(input('Guess a number between 1 and 10:'))
    attempts-=1
    
    if not 1<=guess<=10:
        print('Your guess is out of range. Please guess a number between 1 and 10.')
        continue
        
    if guess < secret_number:
        message = 'Too low.'
    
    elif guess > secret_number:
        message = 'Too high.'
    
    else:
        print()
        print('Congratulations! You guessed the correct number.')
        break

    if attempts > 0:
        print(message, 'Try again.')
    else:
        print(message)
    print()

    if guess != secret_number and attempts > 0:
        print(f'Attempts remaining: {attempts}')
   

else:
    print('The secret number was:',secret_number)
    print('Game over! Better luck next time!')
```

### **Multiplication Table Generator**

**Problem Statement:** Create a Python program that generates and prints a multiplication table (from 1 to 10) for a given number using a for loop and the range function.

```python
num = int(input('Enter the number for which you want the multiplication table:'))
print()
p=1

for i in range(1,11):
    p=num*i
    print(f'{num} x {i} = {p}')
    i+=1
```

## **BMI Calculator**

**Problem Statement:** Create a Python program that calculates the Body Mass Index (BMI).

```python
def calculate_bmi(weight, height):
    bmi = weight/height**2
    print()
    print(f'Your BMI is: {bmi:.2f}')

while True:
    weight = float(input('Enter your weight in kg:'))
    
    if weight>0:
        while True:
            height = float(input('Enter your height in meters'))  
            
            if height>0:
                calculate_bmi(weight,height)
                break
            
            else:
                print('Invalid Height!')
                print()
                continue
            
        break
    
    else:
        print('Invalid Weight!')
        print()
        continue
```
