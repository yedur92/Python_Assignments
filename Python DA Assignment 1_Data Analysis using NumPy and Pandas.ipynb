## **Numpy Array Operations:**

```python
import numpy as np
import pandas as pd
```

**Problem Statement**: To learn how to use NumPy for creating arrays, performing basic
array operations, and conducting numerical computations through analyzing
temperature data.

**Scenario**: You are given a list of daily average temperatures (in degrees Celsius) recorded over two weeks. Your task is to use NumPy to analyze this data.

**1. Create a 1D NumPy Array:**

```python
temperatures_w1 = np.array([22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9])
temperatures_w1
```

**2. Inspection and Properties:**

```python
np.info(temperatures_w1)
print()
print('Shape of temperatures_w1 array :',temperatures_w1.shape)
print('Data Type of temperatures_w1 atrray :',temperatures_w1.dtype)
print('Size of temperatures_w1 array :',(temperatures_w1.size))
```

**3. Array Operations:**

```python
temperatures_in_F_w1 = temperatures_w1 * (9/5) + 32
print('The temperatures for Week 1 in Fahrenheit are :',temperatures_in_F_w1)
print()
print('The maximum temperature of Week 1 was :', temperatures_in_F_w1.max())
print()
print('The minimum temperature of Week 1 was :', temperatures_in_F_w1.min())
print()
print(f'The mean temperature of Week 1 was : {temperatures_in_F_w1.mean():.2f}')
```

**4. Array Slicing and Indexing:**

```python
print('The temperatures for the first three days of the week :',temperatures_in_F_w1[:3])
print()
print('The temperatures for the weekend are :',temperatures_in_F_w1[-2:])
print()
print('The temperatures for the middle three days of the week are :',temperatures_in_F_w1[2:5])
```

**5. Create a 2D Array:**

```python
temperatures = np.array([[22.5, 25.3, 20.8, 23.4, 26.1, 24.8, 21.9],[19.2, 22.5, 21.3, 24.0, 23.5, 22.8, 20.1]])
print(temperatures)
```

**6. Inspect and Slice the 2D Array:**

```python
np.info(temperatures)
print()
print('Shape of temperatures array :',temperatures.shape)
print('Data Type of temperatures_w1 :',temperatures.dtype)
print('Size of temperatures_w1 :',(temperatures.size))
```

```python
print('The temperatures for week 1 are :', temperatures[0])
print()
print('The temperatures for first weekend are :', temperatures[0][-2:])
print()
print('The temperatures for second weekend are :', temperatures[1][-2:])
```

## **Pandas Series:**

**1. Creating Pandas Series:**

```python
marks = pd.Series([95, 92, 89, 85, 80],['Rank1', 'Rank2', 'Rank3', 'Rank4', 'Rank5'])
marks
```

**2. Indexing and Slicing:**

```python
print('The mark of the 1st rank student is :',marks.iloc[0])
```

```python
print('The marks of the top 3 ranked students are :')
marks.loc['Rank1':'Rank3']
```

```python
print('The mark of the 3rd rank student is',marks.iloc[2])
```

```python
print('The ranks of students with marks greater than 90 :')
marks[marks>90]
```

**3. Manipulating Series:**

```python
marks['Rank1'] = 100
marks
```

```python
new_marks = marks.drop('Rank5')
# new_marks = marks.drop(marks.index[-1])
new_marks
```

```python
CGPA = new_marks/10
CGPA
```
